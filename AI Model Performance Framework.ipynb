# PREPROCESSING

```python
# ============================================================
# STEP 4: MIN-MAX NORMALIZATION + SMOTE
# ============================================================

import pandas as pd
import numpy as np
import os
import joblib

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from imblearn.over_sampling import SMOTE
from sklearn.neighbors import NearestNeighbors


# ------------------------------------------------------------
# 1. File paths
# ------------------------------------------------------------

input_dir = r"E:\Projects\1_ai_model_performance\Dataset\split_data"

train_path = os.path.join(
    input_dir,
    "heterogeneous_ai_model_performance_train.csv"
)

test_path = os.path.join(
    input_dir,
    "heterogeneous_ai_model_performance_test.csv"
)


# ------------------------------------------------------------
# 2. Load train and test datasets
# ------------------------------------------------------------

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)


# ------------------------------------------------------------
# 3. Fixed columns
# ------------------------------------------------------------

domain_column = "Domain"
problem_column = "Problem_Type"
target_column = "Best_Model_Label"


# ------------------------------------------------------------
# 4. Columns excluded from RFMC predictors
# ------------------------------------------------------------

excluded_columns = [
    "SMOTE_Applied",
    "PCA_Components",
    "Dimensionality_Reduction_Ratio",
    "Optimization_Status",
    "Bayesian_Iteration_Count",
    "Hyperparameter_Score",
    "Convergence_Score",
    "Model_Name",
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score",
    "ROC_AUC",
    "Training_Time_sec",
    "Prediction_Time_ms",
    "Memory_Usage_MB",
    "Robustness_Score",
    "Computational_Efficiency",
    "Overall_Performance_Score"
]


# ------------------------------------------------------------
# 5. Target
# ------------------------------------------------------------

y_train = train_df[target_column].copy()
y_test = test_df[target_column].copy()


# ------------------------------------------------------------
# 6. Keep Domain and Problem_Type unchanged
# ------------------------------------------------------------

domain_train = train_df[domain_column].copy()
problem_train = train_df[problem_column].copy()

domain_test = test_df[domain_column].copy()
problem_test = test_df[problem_column].copy()


# ------------------------------------------------------------
# 7. Select predictor columns
# ------------------------------------------------------------

predictor_columns = [
    col for col in train_df.columns
    if col not in excluded_columns
    and col not in [
        domain_column,
        problem_column,
        target_column
    ]
]


X_train = train_df[predictor_columns].copy()
X_test = test_df[predictor_columns].copy()


# ------------------------------------------------------------
# 8. Identify numerical and categorical predictors
# ------------------------------------------------------------

numerical_columns = X_train.select_dtypes(
    include=[
        "int64",
        "float64",
        "int32",
        "float32"
    ]
).columns.tolist()

categorical_columns = X_train.select_dtypes(
    include=[
        "object",
        "category",
        "bool"
    ]
).columns.tolist()


# ------------------------------------------------------------
# 9. Numerical preprocessing
# ------------------------------------------------------------

numerical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            MinMaxScaler()
        )
    ]
)


# ------------------------------------------------------------
# 10. Categorical preprocessing
# ------------------------------------------------------------

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)


# ------------------------------------------------------------
# 11. Combined preprocessing
# ------------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numerical_pipeline,
            numerical_columns
        ),
        (
            "cat",
            categorical_pipeline,
            categorical_columns
        )
    ],
    remainder="drop"
)


# ------------------------------------------------------------
# 12. FIT preprocessing ONLY on training data
# ------------------------------------------------------------

X_train_processed = preprocessor.fit_transform(
    X_train
)


# ------------------------------------------------------------
# 13. Transform test data
# ------------------------------------------------------------

X_test_processed = preprocessor.transform(
    X_test
)


# ------------------------------------------------------------
# 14. Convert to NumPy
# ------------------------------------------------------------

X_train_processed = np.asarray(
    X_train_processed,
    dtype=np.float32
)

X_test_processed = np.asarray(
    X_test_processed,
    dtype=np.float32
)


# ============================================================
# 15. SMOTE — TRAINING DATA ONLY
# ============================================================

smote = SMOTE(
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_processed,
    y_train
)


# ------------------------------------------------------------
# 16. Convert SMOTE output to DataFrame
# ------------------------------------------------------------

feature_names = preprocessor.get_feature_names_out()


# ------------------------------------------------------------
# 17. Short feature names
# ------------------------------------------------------------

short_names = []

for name in feature_names:

    name = name.replace("num__", "")
    name = name.replace("cat__", "")

    replacements = {
        "Sample_Size": "SampleSize",
        "Feature_Count": "FeatureCount",
        "Numerical_Features": "NumFeatures",
        "Categorical_Features": "CatFeatures",
        "Missing_Value_Rate": "MissingRate",
        "Imbalance_Ratio": "Imbalance",
        "Noise_Level": "Noise",
        "Outlier_Rate": "OutlierRate",
        "Feature_Correlation_Score": "CorrScore",
        "Class_Overlap_Score": "OverlapScore",
        "Data_Complexity_Level_": "Complexity_",
        "Data_Complexity_Score": "ComplexityScore",
        "Low": "Low",
        "Medium": "Med",
        "High": "High"
    }

    for old, new in replacements.items():
        name = name.replace(old, new)

    short_names.append(name)


# ------------------------------------------------------------
# 18. Create SMOTE training DataFrame
# ------------------------------------------------------------

train_smote_features = pd.DataFrame(
    X_train_smote,
    columns=short_names
)


# ============================================================
# 19. Preserve Domain and Problem_Type
# ============================================================
# SMOTE creates new training feature vectors.
# Assign metadata using the nearest original training
# observation in the normalized feature space.

nearest_model = NearestNeighbors(
    n_neighbors=1
)

nearest_model.fit(
    X_train_processed
)

nearest_indices = nearest_model.kneighbors(
    X_train_smote,
    return_distance=False
).ravel()


domain_smote = domain_train.iloc[
    nearest_indices
].reset_index(drop=True)

problem_smote = problem_train.iloc[
    nearest_indices
].reset_index(drop=True)


# ------------------------------------------------------------
# 20. Add Domain as FIRST column
# ------------------------------------------------------------

train_smote_features.insert(
    0,
    domain_column,
    domain_smote.to_numpy()
)


# ------------------------------------------------------------
# 21. Add Problem_Type as SECOND column
# ------------------------------------------------------------

train_smote_features.insert(
    1,
    problem_column,
    problem_smote.to_numpy()
)


# ------------------------------------------------------------
# 22. Add Best_Model_Label as LAST column
# ------------------------------------------------------------

train_smote_features[target_column] = (
    np.asarray(y_train_smote)
)


# ============================================================
# 23. Create TEST dataset
# ============================================================
# Test data is only transformed.
# NO SMOTE is applied to test data.

test_features_df = pd.DataFrame(
    X_test_processed,
    columns=short_names
)


# ------------------------------------------------------------
# 24. Add Domain as FIRST column
# ------------------------------------------------------------

test_features_df.insert(
    0,
    domain_column,
    domain_test.to_numpy()
)


# ------------------------------------------------------------
# 25. Add Problem_Type as SECOND column
# ------------------------------------------------------------

test_features_df.insert(
    1,
    problem_column,
    problem_test.to_numpy()
)


# ------------------------------------------------------------
# 26. Add Best_Model_Label as LAST column
# ------------------------------------------------------------

test_features_df[target_column] = (
    y_test.to_numpy()
)


# ============================================================
# 27. Create output directory
# ============================================================

processed_dir = (
    r"E:\Projects\1_ai_model_performance\Dataset\processed_data"
)

os.makedirs(
    processed_dir,
    exist_ok=True
)


# ============================================================
# 28. Save SMOTE training dataset
# ============================================================

smote_train_path = os.path.join(
    processed_dir,
    "heterogeneous_ai_model_performance_train_SMOTE.csv"
)

train_smote_features.to_csv(
    smote_train_path,
    index=False
)


# ============================================================
# 29. Save normalized test dataset
# ============================================================

normalized_test_path = os.path.join(
    processed_dir,
    "heterogeneous_ai_model_performance_test_normalized.csv"
)

test_features_df.to_csv(
    normalized_test_path,
    index=False
)


# ============================================================
# 30. Save preprocessing pipeline
# ============================================================

preprocessor_path = os.path.join(
    processed_dir,
    "training_preprocessor.pkl"
)

joblib.dump(
    preprocessor,
    preprocessor_path
)


# ============================================================
# 31. Save SMOTE object
# ============================================================

smote_path = os.path.join(
    processed_dir,
    "SMOTE_model.pkl"
)

joblib.dump(
    smote,
    smote_path
)


# ============================================================
# OUTPUT ONLY
# ============================================================

print("=" * 70)
print("ORIGINAL DATASET SHAPE")
print("=" * 70)

print(
    "Original Train Shape:",
    train_df.shape
)

print(
    "Original Test Shape :",
    test_df.shape
)


print("\n" + "=" * 70)
print("MIN-MAX + SMOTE DATASET SHAPE")
print("=" * 70)

print(
    "SMOTE Train Shape:",
    train_smote_features.shape
)

print(
    "Test Shape:",
    test_features_df.shape
)

print("\n" + "=" * 60)
print("CLASS DISTRIBUTION BEFORE SMOTE")
print("=" * 60)

print(train_df["Best_Model_Label"].value_counts())



print("\n" + "=" * 70)
print("TRAIN CLASS DISTRIBUTION AFTER SMOTE")
print("=" * 70)

print(
    train_smote_features[
        target_column
    ].value_counts()
)


print("\n" + "=" * 70)
print("SMOTE TRAIN DATA - HEAD")
print("=" * 70)

print(
    train_smote_features.head()
)


print("\n" + "=" * 70)
print("TEST DATA - HEAD")
print("=" * 70)

print(
    test_features_df.head()
)
```

# FEATURE EXTRACTION

```python
# ============================================================
# STEP 5: PCA DIMENSIONALITY REDUCTION
# ============================================================

import os
import joblib
import pandas as pd
import numpy as np

from sklearn.decomposition import PCA

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"

TRAIN_INPUT = os.path.join(
    BASE_DIR,"processed_data","heterogeneous_ai_model_performance_train_SMOTE.csv"
)

TEST_INPUT = os.path.join(
    BASE_DIR,
    "processed_data",
    "heterogeneous_ai_model_performance_test_normalized.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Feature_Extracted"
)

TRAIN_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "heterogeneous_ai_model_performance_train_PCA.csv"
)

TEST_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "heterogeneous_ai_model_performance_test_PCA.csv"
)

PCA_MODEL_PATH = os.path.join(
    OUTPUT_DIR,
    "PCA_model.pkl"
)

# ============================================================
# LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_INPUT)
test_df = pd.read_csv(TEST_INPUT)

print("=" * 70)
print("PCA DIMENSIONALITY REDUCTION")
print("=" * 70)

print("\nInput Training Shape :", train_df.shape)
print("Input Testing Shape  :", test_df.shape)

# ============================================================
# FIXED COLUMNS
# ============================================================

DOMAIN_COL = "Domain"
PROBLEM_COL = "Problem_Type"
TARGET_COL = "Best_Model_Label"

# ============================================================
# CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [
    DOMAIN_COL,
    PROBLEM_COL,
    TARGET_COL
]

for col in required_columns:
    if col not in train_df.columns:
        raise ValueError(f"Missing column in training data: {col}")

    if col not in test_df.columns:
        raise ValueError(f"Missing column in testing data: {col}")

# ============================================================
# PRESERVE DOMAIN AND PROBLEM TYPE
# ============================================================

train_domain = train_df[DOMAIN_COL].reset_index(drop=True)
train_problem = train_df[PROBLEM_COL].reset_index(drop=True)
train_target = train_df[TARGET_COL].reset_index(drop=True)

test_domain = test_df[DOMAIN_COL].reset_index(drop=True)
test_problem = test_df[PROBLEM_COL].reset_index(drop=True)
test_target = test_df[TARGET_COL].reset_index(drop=True)

# ============================================================
# SELECT PCA INPUT FEATURES
# ============================================================
# Domain, Problem_Type and Best_Model_Label are NOT included
# in PCA.
#
# All remaining columns in the Step-4 processed datasets are
# already numerical/model-ready features.
# ============================================================

pca_features = [
    col for col in train_df.columns
    if col not in [
        DOMAIN_COL,
        PROBLEM_COL,
        TARGET_COL
    ]
]

# Make sure train and test have identical PCA feature columns
missing_test_features = [
    col for col in pca_features
    if col not in test_df.columns
]

if missing_test_features:
    raise ValueError(
        f"Test data is missing PCA features: {missing_test_features}"
    )

print("\nNumber of PCA input features:", len(pca_features))

print("\nPCA Input Features:")
for i, col in enumerate(pca_features, start=1):
    print(f"{i:02d}. {col}")

# ============================================================
# CREATE PCA MATRICES
# ============================================================

X_train = train_df[pca_features].copy()
X_test = test_df[pca_features].copy()

# Ensure numerical data
X_train = X_train.apply(pd.to_numeric, errors="coerce")
X_test = X_test.apply(pd.to_numeric, errors="coerce")

# Check missing values
if X_train.isnull().sum().sum() > 0:
    raise ValueError(
        "Missing/invalid values detected in PCA training features."
    )

if X_test.isnull().sum().sum() > 0:
    raise ValueError(
        "Missing/invalid values detected in PCA testing features."
    )

# ============================================================
# PCA
# ============================================================
# Retain 95% cumulative explained variance.
#
# IMPORTANT:
# PCA is fitted ONLY on the SMOTE-balanced training data.
# The test data is transformed using the same PCA model.
# ============================================================

pca = PCA(
    n_components=0.95,
    svd_solver="full"
)

# FIT PCA ONLY ON TRAINING DATA
X_train_pca = pca.fit_transform(X_train)

# TRANSFORM TEST DATA USING THE TRAINED PCA
X_test_pca = pca.transform(X_test)

# ============================================================
# PCA INFORMATION
# ============================================================

n_components = pca.n_components_
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("\n" + "=" * 70)
print("PCA RESULTS")
print("=" * 70)

print(f"Original number of features : {X_train.shape[1]}")
print(f"Selected PCA components     : {n_components}")
print(
    f"Explained variance retained : "
    f"{cumulative_variance[-1] * 100:.2f}%"
)

# ============================================================
# CREATE PCA DATAFRAMES
# ============================================================

pca_columns = [
    f"PCA_{i}"
    for i in range(1, n_components + 1)
]

train_pca_df = pd.DataFrame(
    X_train_pca,
    columns=pca_columns
)

test_pca_df = pd.DataFrame(
    X_test_pca,
    columns=pca_columns
)

# ============================================================
# ADD DOMAIN AND PROBLEM TYPE
# ============================================================
# Required column order:
#
# 1. Domain
# 2. Problem_Type
# 3. PCA_1 ... PCA_n
# Last. Best_Model_Label
# ============================================================

train_final = pd.concat(
    [
        train_domain,
        train_problem,
        train_pca_df,
        train_target
    ],
    axis=1
)

test_final = pd.concat(
    [
        test_domain,
        test_problem,
        test_pca_df,
        test_target
    ],
    axis=1
)

# Explicitly enforce column order
train_final = train_final[
    [DOMAIN_COL, PROBLEM_COL] +
    pca_columns +
    [TARGET_COL]
]

test_final = test_final[
    [DOMAIN_COL, PROBLEM_COL] +
    pca_columns +
    [TARGET_COL]
]

# ============================================================
# SAVE PCA DATA
# ============================================================

train_final.to_csv(
    TRAIN_OUTPUT,
    index=False
)

test_final.to_csv(
    TEST_OUTPUT,
    index=False
)

# ============================================================
# SAVE PCA MODEL
# ============================================================

joblib.dump(
    pca,
    PCA_MODEL_PATH
)

# ============================================================
# FINAL OUTPUT INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL PCA DATA")
print("=" * 70)

print("PCA Training Shape :", train_final.shape)
print("PCA Testing Shape  :", test_final.shape)

print("\nTraining Class Distribution:")
print(
    train_final[TARGET_COL].value_counts()
)

print("\nTesting Class Distribution:")
print(
    test_final[TARGET_COL].value_counts()
)

print("\nTrain rows:")
print(train_final.head())

print("\nTest Rows:")
print(test_final.head())

print("\nSaved Files:")
print(TRAIN_OUTPUT)
print(TEST_OUTPUT)
print(PCA_MODEL_PATH)
```

```python
# ============================================================
# STEP 6–8:
# RFMC + BAYESIAN OPTIMIZATION + BO-RFMC
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd
import optuna

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"
FE_DIR = os.path.join(BASE_DIR, "Feature_Extracted")

TRAIN_PCA_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_train_PCA.csv"
)

TEST_PCA_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_test_PCA.csv"
)

RFMC_MODEL_PATH = os.path.join(
    FE_DIR,
    "RFMC_model.pkl"
)

BO_RFMC_MODEL_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_model.pkl"
)

BO_RESULTS_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_trials.csv"
)

CV_RESULTS_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_5Fold_CV_Results.csv"
)

EVALUATION_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_evaluation_data.csv"
)

# ============================================================
# LOAD STEP 5 DATA
# ============================================================

train_df = pd.read_csv(TRAIN_PCA_PATH)
test_df = pd.read_csv(TEST_PCA_PATH)

print("=" * 80)
print("STEP 6–8: RFMC + BAYESIAN OPTIMIZATION + BO-RFMC")
print("=" * 80)

print("\nTraining Shape :", train_df.shape)
print("Testing Shape  :", test_df.shape)

# ============================================================
# COLUMN DEFINITIONS
# ============================================================

DOMAIN_COL = "Domain"
TASK_COL = "Problem_Type"
TARGET_COL = "Best_Model_Label"

# ============================================================
# PCA FEATURES
# ============================================================

PCA_FEATURES = [
    col for col in train_df.columns
    if col.startswith("PCA_")
]

print("\nNumber of PCA Features:", len(PCA_FEATURES))
print("PCA Features:")
print(PCA_FEATURES)

# ============================================================
# INPUT / TARGET
# ============================================================

X_train = train_df[PCA_FEATURES].copy()
y_train = train_df[TARGET_COL].copy()

X_test = test_df[PCA_FEATURES].copy()
y_test = test_df[TARGET_COL].copy()

# ============================================================
# TARGET DISTRIBUTION
# ============================================================

class_labels = list(y_train.unique())

print("\nTarget Classes:")

for cls in class_labels:
    print(
        f"{cls:<25} : "
        f"{(y_train == cls).sum()}"
    )

# ============================================================
# STEP 6
# RFMC META-CLASSIFIER
# ============================================================

print("\n" + "=" * 80)
print("RFMC META-CLASSIFIER")
print("=" * 80)

print("\nTraining RFMC...")

rfmc_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=20,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

rfmc_model.fit(
    X_train,
    y_train
)

# Save RFMC
joblib.dump(
    rfmc_model,
    RFMC_MODEL_PATH
)

print("\nRFMC saved:")
print(RFMC_MODEL_PATH)

# ============================================================
# STEP 7
# BAYESIAN OPTIMIZATION
# ============================================================

print("\n" + "=" * 80)
print("BAYESIAN OPTIMIZATION")
print("=" * 80)

print("\nStarting Bayesian Optimization")

# ============================================================
# 5-FOLD INTERNAL CV
# ============================================================

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("Optimization trials : 5")
print("Internal CV         : 5-Fold")
print("Objective            : Mean CV Accuracy")

# ============================================================
# OPTUNA OBJECTIVE
# ============================================================

def objective(trial):

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators",
            50,
            500
        ),

        "max_depth": trial.suggest_int(
            "max_depth",
            5,
            50
        ),

        "min_samples_split": trial.suggest_int(
            "min_samples_split",
            2,
            20
        ),

        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf",
            1,
            10
        ),

        "max_features": trial.suggest_categorical(
            "max_features",
            ["sqrt", "log2"]
        ),

        "bootstrap": trial.suggest_categorical(
            "bootstrap",
            [True, False]
        )
    }

    model = RandomForestClassifier(
        **params,
        random_state=42,
        n_jobs=-1
    )

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="accuracy",
        n_jobs=-1,
        return_train_score=False
    )

    mean_accuracy = scores["test_score"].mean()

    return mean_accuracy


# ============================================================
# CREATE OPTUNA STUDY
# ============================================================

sampler = optuna.samplers.TPESampler(
    seed=42
)

study = optuna.create_study(
    direction="maximize",
    sampler=sampler
)

# ============================================================
# RUN BAYESIAN OPTIMIZATION
# ============================================================

study.optimize(
    objective,
    n_trials=5,
    show_progress_bar=True
)

# ============================================================
# SAVE BO TRIAL RESULTS
# ============================================================

bo_trials_df = study.trials_dataframe()

bo_trials_df.to_csv(
    BO_RESULTS_PATH,
    index=False
)

print("\nBO trial results saved:")
print(BO_RESULTS_PATH)

# ============================================================
# BEST PARAMETERS
# ============================================================

best_params = study.best_params
best_cv_accuracy = study.best_value

# ============================================================
# STEP 8
# OPTIMIZED BO-RFMC
# ============================================================

print("\n" + "=" * 80)
print("OPTIMIZED BO-RFMC")
print("=" * 80)

print("\nTraining final BO-RFMC...")

bo_rfmc_model = RandomForestClassifier(
    **best_params,
    random_state=42,
    n_jobs=-1
)

# Train on complete SMOTE + PCA training data
bo_rfmc_model.fit(
    X_train,
    y_train
)

print("BO-RFMC training completed.")

# ============================================================
# SAVE FINAL BO-RFMC MODEL
# ============================================================

joblib.dump(
    bo_rfmc_model,
    BO_RFMC_MODEL_PATH
)

# ============================================================
# 5-FOLD CV OF SELECTED BO-RFMC
# ============================================================

selected_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_validate(
    bo_rfmc_model,
    X_train,
    y_train,
    cv=selected_cv,
    scoring={
        "accuracy": "accuracy",
        "precision": "precision_macro",
        "recall": "recall_macro",
        "f1": "f1_macro"
    },
    n_jobs=-1,
    return_train_score=False
)

# ============================================================
# CREATE 5-FOLD RESULTS
# ============================================================

fold_results = pd.DataFrame({
    "Fold": np.arange(1, 6),

    "Accuracy": cv_scores["test_accuracy"],

    "Precision": cv_scores["test_precision"],

    "Recall": cv_scores["test_recall"],

    "F1": cv_scores["test_f1"]
})

mean_row = pd.DataFrame({
    "Fold": ["Mean"],

    "Accuracy": [
        fold_results["Accuracy"].mean()
    ],

    "Precision": [
        fold_results["Precision"].mean()
    ],

    "Recall": [
        fold_results["Recall"].mean()
    ],

    "F1": [
        fold_results["F1"].mean()
    ]
})

fold_results_final = pd.concat(
    [
        fold_results,
        mean_row
    ],
    ignore_index=True
)

# ============================================================
# SAVE 5-FOLD RESULTS
# ============================================================

fold_results_final.to_csv(
    CV_RESULTS_PATH,
    index=False
)

# ============================================================
# PREPARE TEST PREDICTIONS
# ============================================================

bo_pred = bo_rfmc_model.predict(
    X_test
)

bo_prob = bo_rfmc_model.predict_proba(
    X_test
)

# ============================================================
# PREPARE EVALUATION DATA
# ============================================================
# This file will be used by the NEXT block.
# ============================================================

evaluation_df = test_df[
    [
        DOMAIN_COL,
        TASK_COL,
        TARGET_COL
    ]
].copy()

evaluation_df["Predicted_Model"] = bo_pred

evaluation_df["Correct"] = (
    evaluation_df[TARGET_COL]
    ==
    evaluation_df["Predicted_Model"]
)

# ============================================================
# SAVE PREDICTED PROBABILITIES
# ============================================================

for i, class_name in enumerate(
    bo_rfmc_model.classes_
):

    safe_name = str(class_name).replace(
        " ",
        "_"
    )

    evaluation_df[
        f"Probability_{safe_name}"
    ] = bo_prob[:, i]

# ============================================================
# SAVE EVALUATION DATA
# ============================================================

evaluation_df.to_csv(
    EVALUATION_PATH,
    index=False
)

# ============================================================
# STORE VARIABLES FOR NEXT BLOCK
# ============================================================

FINAL_MODEL = bo_rfmc_model

FINAL_PREDICTIONS = bo_pred

FINAL_PROBABILITIES = bo_prob

FINAL_Y_TEST = y_test.copy()

FINAL_EVALUATION_DF = evaluation_df.copy()

FINAL_FOLD_RESULTS = fold_results_final.copy()

FINAL_BEST_PARAMS = best_params.copy()

FINAL_BO_CV_ACCURACY = best_cv_accuracy

# ============================================================
# FINAL FILE OUTPUT ONLY
# ============================================================

print("\n")
print("Files generated:")
print("1.", RFMC_MODEL_PATH)
print("2.", BO_RFMC_MODEL_PATH)
print("3.", BO_RESULTS_PATH)
print("4.", CV_RESULTS_PATH)
print("5.", EVALUATION_PATH)

print("\n" + "=" * 80)
```

```python
# ============================================================
# STEP 9: BO-RFMC FIVE-FOLD CV + FINAL EVALUATION
# ============================================================

import os
import numpy as np
import pandas as pd
import joblib

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)


# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"

FE_DIR = os.path.join(
    BASE_DIR,
    "Feature_Extracted"
)

SPLIT_DIR = os.path.join(
    BASE_DIR,
    "split_data"
)

MODEL_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_model.pkl"
)

CV_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_5Fold_CV_Results.csv"
)

EVAL_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_evaluation_data.csv"
)

ORIGINAL_TEST_PATH = os.path.join(
    SPLIT_DIR,
    "heterogeneous_ai_model_performance_test.csv"
)

OUTPUT_PATH = os.path.join(
    FE_DIR,
    "BO_RFMC_Task_Wise_Evaluation.csv"
)


# ============================================================
# LOAD FILES
# ============================================================

bo_rfmc_model = joblib.load(
    MODEL_PATH
)

cv_results = pd.read_csv(
    CV_PATH
)

eval_df = pd.read_csv(
    EVAL_PATH
)

original_test = pd.read_csv(
    ORIGINAL_TEST_PATH
)


# ============================================================
# PART 1: FIVE-FOLD CROSS-VALIDATION
# ============================================================

print("\n")
print("=" * 110)
print("BO-RFMC 5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 110)

cv_display = cv_results.copy()

cv_columns = [
    "Fold",
    "Accuracy",
    "Precision",
    "Recall",
    "F1"
]

cv_columns = [
    col
    for col in cv_columns
    if col in cv_display.columns
]

cv_display = cv_display[
    cv_columns
].copy()


# ============================================================
# FORMAT CV VALUES
# ============================================================

for col in cv_display.columns:

    if col != "Fold":

        cv_display[col] = cv_display[col].apply(
            lambda x:
            f"{x*100:.2f}"
            if pd.notna(x)
            else "NaN"
        )


# ============================================================
# LEFT-ALIGNED CV TABLE
# ============================================================

cv_display = cv_display.astype(str)

cv_widths = {}

for col in cv_display.columns:

    cv_widths[col] = max(
        len(col),
        cv_display[col].map(len).max()
    )


print(
    "  ".join(
        col.ljust(cv_widths[col])
        for col in cv_display.columns
    )
)

print(
    "  ".join(
        "-" * cv_widths[col]
        for col in cv_display.columns
    )
)

for _, row in cv_display.iterrows():

    print(
        "  ".join(
            str(row[col]).ljust(
                cv_widths[col]
            )
            for col in cv_display.columns
        )
    )


# ============================================================
# ADD ORIGINAL TEST PERFORMANCE INFORMATION
# ============================================================

# ------------------------------------------------------------
# Robustness Score
# ------------------------------------------------------------

if "Robustness_Score" in original_test.columns:

    eval_df["Robustness_Score"] = pd.to_numeric(
        original_test[
            "Robustness_Score"
        ],
        errors="coerce"
    ).values

else:

    eval_df["Robustness_Score"] = np.nan


# ------------------------------------------------------------
# Computational Efficiency
# ------------------------------------------------------------

if "Computational_Efficiency" in original_test.columns:

    eval_df["Computational_Efficiency"] = pd.to_numeric(
        original_test[
            "Computational_Efficiency"
        ],
        errors="coerce"
    ).values

else:

    eval_df["Computational_Efficiency"] = np.nan


# ------------------------------------------------------------
# Overall Performance Score
# ------------------------------------------------------------

if "Overall_Performance_Score" in original_test.columns:

    eval_df["Overall_Performance_Score"] = pd.to_numeric(
        original_test[
            "Overall_Performance_Score"
        ],
        errors="coerce"
    ).values

else:

    eval_df["Overall_Performance_Score"] = np.nan


# ============================================================
# MODEL CLASSES
# ============================================================

model_classes = list(
    bo_rfmc_model.classes_
)

probability_columns = [
    f"Probability_{cls}"
    for cls in model_classes
    if f"Probability_{cls}" in eval_df.columns
]


# ============================================================
# METRIC FUNCTION
# ============================================================

def calculate_metrics(data):

    # --------------------------------------------------------
    # TRUE / PREDICTED LABELS
    # --------------------------------------------------------

    y_true = data[
        "Best_Model_Label"
    ].astype(str)

    y_pred = data[
        "Predicted_Model"
    ].astype(str)


    # --------------------------------------------------------
    # ACCURACY
    # --------------------------------------------------------

    acc = accuracy_score(
        y_true,
        y_pred
    )


    # --------------------------------------------------------
    # PRECISION
    # --------------------------------------------------------

    pre = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # RECALL
    # --------------------------------------------------------

    rec = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # F1 SCORE
    # --------------------------------------------------------

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )


    # --------------------------------------------------------
    # ROC-AUC
    # --------------------------------------------------------

    roc_auc = np.nan

    if len(probability_columns) >= 2:

        try:

            y_prob = data[
                probability_columns
            ].astype(float).values

            present_classes = list(
                pd.unique(y_true)
            )

            # Multiclass ROC-AUC
            if len(present_classes) > 2:

                roc_auc = roc_auc_score(
                    y_true,
                    y_prob,
                    multi_class="ovr",
                    average="macro",
                    labels=model_classes
                )

            # Binary ROC-AUC
            elif len(present_classes) == 2:

                positive_class = sorted(
                    present_classes
                )[1]

                positive_column = (
                    f"Probability_{positive_class}"
                )

                if positive_column in data.columns:

                    y_binary = (
                        y_true == positive_class
                    ).astype(int)

                    roc_auc = roc_auc_score(
                        y_binary,
                        data[
                            positive_column
                        ].astype(float)
                    )

        except Exception:

            roc_auc = np.nan


    # --------------------------------------------------------
    # ROBUSTNESS
    # --------------------------------------------------------

    robustness = data[
        "Robustness_Score"
    ].mean()


    # --------------------------------------------------------
    # COMPUTATIONAL EFFICIENCY
    # --------------------------------------------------------

    computational_efficiency = data[
        "Computational_Efficiency"
    ].mean()


    # --------------------------------------------------------
    # OVERALL SCORE
    # --------------------------------------------------------

    overall_score = data[
        "Overall_Performance_Score"
    ].mean()


    # --------------------------------------------------------
    # RETURN METRICS
    # --------------------------------------------------------

    return {

        "Acc": acc*100,

        "Pre": pre*100,

        "Rec": rec*100,

        "F1": f1*100,

        "Roc-auc": roc_auc,

        "Robustness":
            robustness,

        "Computational Efficiency":
            computational_efficiency,

        "Overall score":
            overall_score
    }


# ============================================================
# PART 3: TASK-WISE EVALUATION
# ============================================================

results = []


# ============================================================
# 1. BINARY CLASSIFICATION
# ============================================================

binary_df = eval_df[
    eval_df["Problem_Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "binary_classification"
].copy()


if len(binary_df) > 0:

    metrics = calculate_metrics(
        binary_df
    )

    row = {
        "Task Type /":
            "Binary_Classification"
    }

    row.update(metrics)

    results.append(row)


# ============================================================
# 2. MULTICLASS CLASSIFICATION
# ============================================================

multiclass_df = eval_df[
    eval_df["Problem_Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "multiclass_classification"
].copy()


if len(multiclass_df) > 0:

    metrics = calculate_metrics(
        multiclass_df
    )

    row = {
        "Task Type /":
            "Multiclass_Classification"
    }

    row.update(metrics)

    results.append(row)


# ============================================================
# 3. PERFORMANCE CLASSIFICATION
# ============================================================

performance_df = eval_df[
    eval_df["Problem_Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "performance_classification"
].copy()

if len(performance_df) > 0:

    metrics = calculate_metrics(
        performance_df
    )

    row = {
        "Task Type /":
            "Performance_Classification"
    }

    row.update(metrics)

    results.append(row)


# ============================================================
# 4. RISK PREDICTION
# ============================================================

risk_df = eval_df[
    eval_df["Problem_Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    == "risk_prediction"
].copy()

if len(risk_df) > 0:

    metrics = calculate_metrics(
        risk_df
    )

    row = {
        "Task Type /":
            "Risk_Prediction"
    }

    row.update(metrics)

    results.append(row)



# ============================================================
# CREATE FINAL TABLE
# ============================================================

task_table = pd.DataFrame(
    results
)


# ============================================================
# COLUMN ORDER
# ============================================================

task_table = task_table[
    [
        "Task Type /",
        "Acc",
        "Pre",
        "Rec",
        "F1",
        "Roc-auc",
        "Robustness",
        "Computational Efficiency",
        "Overall score"
    ]
]


# ============================================================
# FORMAT NUMERIC VALUES
# ============================================================

numeric_columns = [
    "Acc",
    "Pre",
    "Rec",
    "F1",
    "Roc-auc",
    "Robustness",
    "Computational Efficiency",
    "Overall score"
]

for col in numeric_columns:

    task_table[col] = task_table[col].apply(
        lambda x:
        f"{x:.3f}"
        if pd.notna(x)
        else "NaN"
    )


# ============================================================
# SAVE FINAL TABLE
# ============================================================

task_table.to_csv(
    OUTPUT_PATH,
    index=False
)


# ============================================================
# PRINT FINAL TABLE
# ============================================================

print("\n")
print("=" * 145)
print(
    "BO-RFMC PROPOSED MODEL – TASK-WISE PERFORMANCE"
)
print("=" * 145)

display_table = task_table.astype(str)

column_widths = {}

for col in display_table.columns:

    column_widths[col] = max(
        len(col),
        display_table[col].map(len).max()
    )


# ============================================================
# HEADER
# ============================================================

print(
    "  ".join(
        col.ljust(column_widths[col])
        for col in display_table.columns
    )
)


# ============================================================
# SEPARATOR
# ============================================================

print(
    "  ".join(
        "-" * column_widths[col]
        for col in display_table.columns
    )
)


# ============================================================
# DATA ROWS
# ============================================================

for _, row in display_table.iterrows():

    print(
        "  ".join(
            str(row[col]).ljust(
                column_widths[col]
            )
            for col in display_table.columns
        )
    )


# ============================================================
# END
# ============================================================

print("=" * 145)

print("\nSaved:")
print(OUTPUT_PATH)
```

```python
# ============================================================
# BASELINE 1: AUTOGLUON
# CLEAN TRAINING VERSION
# ============================================================

import os
import warnings
import logging
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# SUPPRESS NON-CRITICAL PYTHON WARNINGS
# ------------------------------------------------------------

warnings.filterwarnings("ignore")

logging.getLogger("autogluon").setLevel(logging.ERROR)
logging.getLogger("lightgbm").setLevel(logging.ERROR)
logging.getLogger("catboost").setLevel(logging.ERROR)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

from autogluon.tabular import TabularPredictor


# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"

FE_DIR = os.path.join(
    BASE_DIR,
    "Feature_Extracted"
)

SPLIT_DIR = os.path.join(
    BASE_DIR,
    "split_data"
)

TRAIN_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_train_PCA.csv"
)

TEST_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_test_PCA.csv"
)

ORIGINAL_TEST_PATH = os.path.join(
    SPLIT_DIR,
    "heterogeneous_ai_model_performance_test.csv"
)

# NEW MODEL DIRECTORY
# Avoids "path already exists" warning
MODEL_PATH = os.path.join(
    FE_DIR,
    "AutoGluon_Baseline_Clean"
)

OUTPUT_PATH = os.path.join(
    FE_DIR,
    "AutoGluon_Task_Wise_Evaluation.csv"
)


# ============================================================
# LOAD DATA
# ============================================================

train_df = pd.read_csv(
    TRAIN_PATH
)

test_df = pd.read_csv(
    TEST_PATH
)

original_test = pd.read_csv(
    ORIGINAL_TEST_PATH
)


# ============================================================
# TARGET AND PCA FEATURES
# ============================================================

TARGET = "Best_Model_Label"

PCA_FEATURES = [
    col
    for col in train_df.columns
    if col.startswith("PCA_")
]


# ============================================================
# TRAIN / TEST DATA
# ============================================================

X_train = train_df[
    PCA_FEATURES
].copy()

y_train = train_df[
    TARGET
].astype(str).copy()

X_test = test_df[
    PCA_FEATURES
].copy()

y_test = test_df[
    TARGET
].astype(str).copy()


# ============================================================
# AUTOGLUON TRAINING DATA
# ============================================================

ag_train = X_train.copy()

ag_train[TARGET] = y_train


# ============================================================
# HEADER
# ============================================================

print("\n")
print("=" * 110)
print("BASELINE 1: AUTOGLUON")
print("=" * 110)

print(
    "\nTraining Shape :",
    X_train.shape
)

print(
    "Testing Shape  :",
    X_test.shape
)


# ============================================================
# TRAIN AUTOGLUON
# ============================================================

print("\nTraining AutoGluon...")


predictor = TabularPredictor(
    label=TARGET,
    problem_type="multiclass",
    eval_metric="accuracy",
    path=MODEL_PATH,
    verbosity=0
)


# ------------------------------------------------------------
# IMPORTANT:
# Only use models that work reliably in the current
# environment.
#
# RF  = Random Forest
# XT  = Extra Trees
# ------------------------------------------------------------

predictor.fit(
    train_data=ag_train,

    hyperparameters={
        "RF": {},
        "XT": {}
    },

    presets="medium_quality",

    time_limit=600,

    num_bag_folds=0,

    num_stack_levels=0
)


# ============================================================
# PREDICTION
# ============================================================

y_pred = predictor.predict(
    X_test
).astype(str)


# ============================================================
# PREDICTION PROBABILITIES
# ============================================================

y_prob_df = predictor.predict_proba(
    X_test
)

model_classes = [
    str(cls)
    for cls in predictor.class_labels
]

y_prob = y_prob_df[
    model_classes
].astype(float).values


# ============================================================
# CREATE EVALUATION DATA
# ============================================================

eval_df = pd.DataFrame()

eval_df["Problem_Type"] = (
    original_test[
        "Problem_Type"
    ]
    .astype(str)
    .values
)

eval_df["Best_Model_Label"] = (
    original_test[
        "Best_Model_Label"
    ]
    .astype(str)
    .values
)

eval_df["Predicted_Model"] = y_pred


# ============================================================
# ADD PROBABILITIES
# ============================================================

for i, cls in enumerate(model_classes):

    eval_df[
        f"Probability_{cls}"
    ] = y_prob[:, i]


# ============================================================
# ADD ORIGINAL TEST PERFORMANCE INFORMATION
# ============================================================

eval_df[
    "Robustness_Score"
] = pd.to_numeric(
    original_test[
        "Robustness_Score"
    ],
    errors="coerce"
).values


eval_df[
    "Computational_Efficiency"
] = pd.to_numeric(
    original_test[
        "Computational_Efficiency"
    ],
    errors="coerce"
).values


eval_df[
    "Overall_Performance_Score"
] = pd.to_numeric(
    original_test[
        "Overall_Performance_Score"
    ],
    errors="coerce"
).values


# ============================================================
# METRIC FUNCTION
# ============================================================

def calculate_metrics(data):

    y_true = data[
        "Best_Model_Label"
    ].astype(str)

    y_pred_local = data[
        "Predicted_Model"
    ].astype(str)


    # Accuracy
    acc = accuracy_score(
        y_true,
        y_pred_local
    )


    # Precision
    pre = precision_score(
        y_true,
        y_pred_local,
        average="macro",
        zero_division=0
    )


    # Recall
    rec = recall_score(
        y_true,
        y_pred_local,
        average="macro",
        zero_division=0
    )


    # F1
    f1 = f1_score(
        y_true,
        y_pred_local,
        average="macro",
        zero_division=0
    )


    # ROC-AUC
    roc_auc = np.nan

    present_classes = list(
        pd.unique(y_true)
    )

    try:

        valid_classes = [
            cls
            for cls in model_classes
            if cls in present_classes
        ]

        if len(valid_classes) > 2:

            valid_columns = [
                f"Probability_{cls}"
                for cls in valid_classes
                if f"Probability_{cls}" in data.columns
            ]

            y_prob_local = data[
                valid_columns
            ].astype(float).values

            roc_auc = roc_auc_score(
                y_true,
                y_prob_local,
                multi_class="ovr",
                average="macro",
                labels=valid_classes
            )

        elif len(valid_classes) == 2:

            positive_class = sorted(
                valid_classes
            )[1]

            positive_column = (
                f"Probability_{positive_class}"
            )

            y_binary = (
                y_true == positive_class
            ).astype(int)

            roc_auc = roc_auc_score(
                y_binary,
                data[
                    positive_column
                ].astype(float)
            )

    except Exception:

        roc_auc = np.nan


    # Robustness
    robustness = data[
        "Robustness_Score"
    ].mean()


    # Computational Efficiency
    computational_efficiency = data[
        "Computational_Efficiency"
    ].mean()


    # Overall Score
    overall_score = data[
        "Overall_Performance_Score"
    ].mean()


    return {

        "Acc": acc*100,

        "Pre": pre*100,

        "Rec": rec*100,

        "F1": f1*100,

        "Roc-auc": roc_auc,

        "Robustness": robustness,

        "Computational Efficiency":
            computational_efficiency,

        "Overall score":
            overall_score
    }


# ============================================================
# TASK-WISE RESULTS
# ============================================================

results = []


# ------------------------------------------------------------
# Binary Classification
# ------------------------------------------------------------

binary_df = eval_df[
    eval_df["Problem_Type"]
    .str.strip()
    .str.lower()
    == "binary_classification"
].copy()

if len(binary_df) > 0:

    row = {
        "Task Type /":
            "Binary_Classification"
    }

    row.update(
        calculate_metrics(binary_df)
    )

    results.append(row)


# ------------------------------------------------------------
# Multiclass Classification
# ------------------------------------------------------------

multiclass_df = eval_df[
    eval_df["Problem_Type"]
    .str.strip()
    .str.lower()
    == "multiclass_classification"
].copy()

if len(multiclass_df) > 0:

    row = {
        "Task Type /":
            "Multiclass_Classification"
    }

    row.update(
        calculate_metrics(multiclass_df)
    )

    results.append(row)


# ------------------------------------------------------------
# Performance Classification
# ------------------------------------------------------------

performance_df = eval_df[
    eval_df["Problem_Type"]
    .str.strip()
    .str.lower()
    == "performance_classification"
].copy()

if len(performance_df) > 0:

    row = {
        "Task Type /":
            "Performance_Classification"
    }

    row.update(
        calculate_metrics(performance_df)
    )

    results.append(row)


# ------------------------------------------------------------
# Risk Prediction
# ------------------------------------------------------------

risk_df = eval_df[
    eval_df["Problem_Type"]
    .str.strip()
    .str.lower()
    == "risk_prediction"
].copy()

if len(risk_df) > 0:

    row = {
        "Task Type /":
            "Risk_Prediction"
    }

    row.update(
        calculate_metrics(risk_df)
    )

    results.append(row)



# ============================================================
# FINAL TABLE
# ============================================================

task_table = pd.DataFrame(
    results
)

task_table = task_table[
    [
        "Task Type /",
        "Acc",
        "Pre",
        "Rec",
        "F1",
        "Roc-auc",
        "Robustness",
        "Computational Efficiency",
        "Overall score"
    ]
]


# ============================================================
# FORMAT VALUES
# ============================================================

numeric_columns = [
    "Acc",
    "Pre",
    "Rec",
    "F1",
    "Roc-auc",
    "Robustness",
    "Computational Efficiency",
    "Overall score"
]

for col in numeric_columns:

    task_table[col] = task_table[col].apply(
        lambda x:
        f"{x:.3f}"
        if pd.notna(x)
        else "NaN"
    )


# ============================================================
# SAVE
# ============================================================

task_table.to_csv(
    OUTPUT_PATH,
    index=False
)


# ============================================================
# PRINT
# ============================================================

print("\n")
print("=" * 145)

print(
    "AUTOGLUON BASELINE – TASK-WISE PERFORMANCE"
)

print("=" * 145)


display_table = task_table.astype(str)

column_widths = {}

for col in display_table.columns:

    column_widths[col] = max(
        len(col),
        display_table[col].map(len).max()
    )


print(
    "  ".join(
        col.ljust(
            column_widths[col]
        )
        for col in display_table.columns
    )
)


print(
    "  ".join(
        "-" * column_widths[col]
        for col in display_table.columns
    )
)


for _, row in display_table.iterrows():

    print(
        "  ".join(
            str(row[col]).ljust(
                column_widths[col]
            )
            for col in display_table.columns
        )
    )


print("=" * 145)


print(
    OUTPUT_PATH
)
```

```python
# ============================================================
# TPOT BASELINE
# TRAIN + TEST + TASK-WISE PERFORMANCE + CSV
# ============================================================

import os
import time
import joblib
import numpy as np
import pandas as pd

from tpot import TPOTClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"

FE_DIR = os.path.join(
    BASE_DIR,
    "Feature_Extracted"
)

TRAIN_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_train_PCA.csv"
)

TEST_PATH = os.path.join(
    FE_DIR,
    "heterogeneous_ai_model_performance_test_PCA.csv"
)

ORIGINAL_TEST_PATH = os.path.join(
    BASE_DIR,
    "split_data",
    "heterogeneous_ai_model_performance_test.csv"
)

MODEL_PATH = os.path.join(
    FE_DIR,
    "TPOT_model.pkl"
)

OUTPUT_CSV = os.path.join(
    FE_DIR,
    "TPOT_Task_Wise_Performance.csv"
)

# ============================================================
# LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_PATH)

test_df = pd.read_csv(TEST_PATH)

original_test_df = pd.read_csv(
    ORIGINAL_TEST_PATH
)

# ============================================================
# PCA FEATURES
# ============================================================

PCA_FEATURES = [
    "PCA_1",
    "PCA_2",
    "PCA_3",
    "PCA_4",
    "PCA_5",
    "PCA_6",
    "PCA_7",
    "PCA_8",
    "PCA_9",
    "PCA_10"
]

TARGET = "Best_Model_Label"

# ============================================================
# TRAIN / TEST DATA
# ============================================================

X_train = train_df[
    PCA_FEATURES
].copy()

y_train = train_df[
    TARGET
].astype(str).copy()

X_test = test_df[
    PCA_FEATURES
].copy()

y_test = test_df[
    TARGET
].astype(str).copy()

# ============================================================
# PRINT ONLY SHAPES
# ============================================================

print("\n" + "=" * 60)
print("TPOT BASELINE")
print("=" * 60)

print(
    "Training shape :",
    train_df.shape
)

print(
    "Testing shape  :",
    test_df.shape
)

# ============================================================
# TPOT CONFIGURATION
# ============================================================

tpot = TPOTClassifier(
    search_space="linear-light",

    # Total search budget
    max_time_mins=2,

    # Maximum time allowed for one candidate
    max_eval_time_mins=1,

    random_state=42,

    # Avoid Dask worker problems
    n_jobs=1,

    verbose=2,

    scatter=False
)

# ============================================================
# TPOT TRAINING
# ============================================================

print("\n" + "=" * 60)
print("TRAINING TPOT")
print("=" * 60)

start_time = time.time()

try:

    tpot.fit(
        X_train,
        y_train
    )

except Exception as e:

    elapsed = time.time() - start_time

    print("\nTPOT training failed.")
    print("Error :", e)
    print(
        f"Elapsed time : {elapsed:.2f} seconds"
    )

    raise

training_time = (
    time.time() - start_time
)

# ============================================================
# SAVE MODEL IMMEDIATELY
# ============================================================

joblib.dump(
    tpot.fitted_pipeline_,
    MODEL_PATH
)

# ============================================================
# TEST PREDICTION
# ============================================================

y_pred_raw = np.asarray(
    tpot.predict(
        X_test
    )
)

# ============================================================
# ORIGINAL CLASS LABELS
# ============================================================

original_labels = sorted(
    y_train.unique()
)

# ============================================================
# TPOT MODEL CLASSES
# ============================================================

model_classes = np.asarray(
    tpot.classes_
)

# ============================================================
# CONVERT TPOT PREDICTIONS
# TO ORIGINAL STRING LABELS
# ============================================================

if np.issubdtype(
    y_pred_raw.dtype,
    np.number
):

    y_pred_numeric = (
        y_pred_raw.astype(int)
    )

    # --------------------------------------------------------
    # Numeric model classes
    # --------------------------------------------------------

    if np.issubdtype(
        model_classes.dtype,
        np.number
    ):

        class_mapping = {}

        for i, class_value in enumerate(
            model_classes
        ):

            class_mapping[
                int(class_value)
            ] = original_labels[i]

        y_pred = np.array([
            class_mapping[int(value)]
            for value in y_pred_numeric
        ])

    # --------------------------------------------------------
    # String model classes
    # --------------------------------------------------------

    else:

        y_pred = np.array([
            original_labels[int(value)]
            for value in y_pred_numeric
        ])

else:

    y_pred = (
        y_pred_raw.astype(str)
    )

# ============================================================
# TRUE LABELS
# ============================================================

y_true = (
    y_test
    .astype(str)
    .values
)

# ============================================================
# PROBABILITY PREDICTIONS
# ============================================================

try:

    y_proba = tpot.predict_proba(
        X_test
    )

except Exception:

    y_proba = None

# ============================================================
# TASK TYPES
# ============================================================

TASK_TYPES = [
    "Binary_Classification",
    "Multiclass_Classification",
    "Performance_Classification",
    "Risk_Prediction"
]

task_results = []

# ============================================================
# TASK-WISE PERFORMANCE
# ============================================================

for task in TASK_TYPES:

    # --------------------------------------------------------
    # Task mask
    # --------------------------------------------------------

    mask = (
        original_test_df[
            "Problem_Type"
        ]
        .astype(str)
        .values
        == task
    )

    # --------------------------------------------------------
    # Skip empty task
    # --------------------------------------------------------

    if mask.sum() == 0:
        continue

    # --------------------------------------------------------
    # Task labels
    # --------------------------------------------------------

    task_y_true = y_true[
        mask
    ]

    task_y_pred = y_pred[
        mask
    ]

    # ========================================================
    # ACCURACY
    # ========================================================

    task_accuracy = accuracy_score(
        task_y_true,
        task_y_pred
    )

    # ========================================================
    # PRECISION
    # ========================================================

    task_precision = precision_score(
        task_y_true,
        task_y_pred,
        average="macro",
        zero_division=0
    )

    # ========================================================
    # RECALL
    # ========================================================

    task_recall = recall_score(
        task_y_true,
        task_y_pred,
        average="macro",
        zero_division=0
    )

    # ========================================================
    # F1-SCORE
    # ========================================================

    task_f1 = f1_score(
        task_y_true,
        task_y_pred,
        average="macro",
        zero_division=0
    )

    # ========================================================
    # ROC-AUC
    # ========================================================

    task_roc_auc = np.nan

    if y_proba is not None:

        try:

            task_proba = y_proba[
                mask
            ]

            # ------------------------------------------------
            # Probability column ordering
            # ------------------------------------------------

            if np.issubdtype(
                model_classes.dtype,
                np.number
            ):

                model_label_order = [
                    original_labels[i]
                    for i in range(
                        len(model_classes)
                    )
                ]

            else:

                model_label_order = [
                    str(c)
                    for c in model_classes
                ]

            # ------------------------------------------------
            # Reorder probabilities
            # ------------------------------------------------

            proba_df = pd.DataFrame(
                task_proba,
                columns=model_label_order
            )

            # Only reorder if all original labels exist
            if all(
                label in proba_df.columns
                for label in original_labels
            ):

                proba_df = proba_df[
                    original_labels
                ]

                task_roc_auc = roc_auc_score(
                    task_y_true,
                    proba_df.values,
                    multi_class="ovr",
                    average="macro",
                    labels=original_labels
                )

        except Exception:

            task_roc_auc = np.nan

    # ========================================================
    # DESCRIPTIVE METRICS
    # ========================================================

    robustness = original_test_df.loc[
        mask,
        "Robustness_Score"
    ].mean()

    computational_efficiency = (
        original_test_df.loc[
            mask,
            "Computational_Efficiency"
        ].mean()
    )

    overall_performance = (
        original_test_df.loc[
            mask,
            "Overall_Performance_Score"
        ].mean()
    )

    # ========================================================
    # STORE RESULT
    # ========================================================

    task_results.append({

        "Task Type":
            task,

        "Accuracy":
            task_accuracy,

        "Precision":
            task_precision,

        "Recall":
            task_recall,

        "F1-Score":
            task_f1,

        "ROC-AUC":
            task_roc_auc,

        "Robustness":
            robustness,

        "Computational Efficiency":
            computational_efficiency,

        "Overall Performance Score":
            overall_performance
    })

# ============================================================
# CREATE RESULT DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    task_results
)

# ============================================================
# SAVE TASK-WISE CSV
# ============================================================

results_df.to_csv(
    OUTPUT_CSV,
    index=False
)

# ============================================================
# PRINT TASK-WISE + OVERALL PERFORMANCE
# ============================================================

print("\n" + "=" * 60)
print("TPOT TASK-WISE PERFORMANCE")
print("=" * 60)

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\nCSV saved:")
print(OUTPUT_CSV)

print("\nModel saved:")
print(MODEL_PATH)
```

# STATISTICAL COMPARISON

```python
import os
import numpy as np
import pandas as pd

from scipy.stats import ttest_rel, t

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset\Feature_Extracted"

AUTOGLUON_PATH = os.path.join(
    BASE_DIR,
    "AutoGluon_Task_Wise_Evaluation.csv"
)

BO_RFMC_PATH = os.path.join(
    BASE_DIR,
    "BO_RFMC_Task_Wise_Evaluation.csv"
)

OUTPUT_PATH = os.path.join(
    BASE_DIR,
    "BO_RFMC_vs_AutoGluon_Statistical_Analysis.csv"
)

autogluon_df = pd.read_csv(
    AUTOGLUON_PATH
)

bo_rfmc_df = pd.read_csv(
    BO_RFMC_PATH
)

TASKS = [
    "Binary_Classification",
    "Multiclass_Classification",
    "Performance_Classification",
    "Risk_Prediction"
]

METRICS = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]


def find_column(df, candidates):

    for column in candidates:

        if column in df.columns:
            return column

    raise KeyError(
        f"None of these columns were found: {candidates}"
    )


AUTO_TASK_COL = find_column(
    autogluon_df,
    ["Task Type", "Task_Type"]
)

BO_TASK_COL = find_column(
    bo_rfmc_df,
    ["Task Type", "Task_Type"]
)


AUTO_METRIC_COLUMNS = {
    "Accuracy": find_column(
        autogluon_df,
        ["Accuracy", "Acc"]
    ),

    "Precision": find_column(
        autogluon_df,
        ["Precision", "Pre"]
    ),

    "Recall": find_column(
        autogluon_df,
        ["Recall", "Rec"]
    ),

    "F1-Score": find_column(
        autogluon_df,
        ["F1-Score", "F1"]
    )
}


BO_METRIC_COLUMNS = {
    "Accuracy": find_column(
        bo_rfmc_df,
        ["Accuracy", "Acc"]
    ),

    "Precision": find_column(
        bo_rfmc_df,
        ["Precision", "Pre"]
    ),

    "Recall": find_column(
        bo_rfmc_df,
        ["Recall", "Rec"]
    ),

    "F1-Score": find_column(
        bo_rfmc_df,
        ["F1-Score", "F1"]
    )
}


auto_filtered = (
    autogluon_df[
        autogluon_df[AUTO_TASK_COL].isin(TASKS)
    ]
    .copy()
)

bo_filtered = (
    bo_rfmc_df[
        bo_rfmc_df[BO_TASK_COL].isin(TASKS)
    ]
    .copy()
)


auto_filtered = (
    auto_filtered
    .set_index(AUTO_TASK_COL)
    .reindex(TASKS)
)

bo_filtered = (
    bo_filtered
    .set_index(BO_TASK_COL)
    .reindex(TASKS)
)


results = []


for metric in METRICS:

    auto_values = (
        auto_filtered[
            AUTO_METRIC_COLUMNS[metric]
        ]
        .astype(float)
        .values
    )

    bo_values = (
        bo_filtered[
            BO_METRIC_COLUMNS[metric]
        ]
        .astype(float)
        .values
    )


    valid = (
        np.isfinite(auto_values)
        &
        np.isfinite(bo_values)
    )

    auto_values = auto_values[valid]
    bo_values = bo_values[valid]


    n = len(auto_values)

    df = n - 1


    auto_mean = np.mean(
        auto_values
    )

    auto_sd = np.std(
        auto_values,
        ddof=1
    )


    bo_mean = np.mean(
        bo_values
    )

    bo_sd = np.std(
        bo_values,
        ddof=1
    )


    differences = (
        bo_values
        -
        auto_values
    )


    mean_difference = np.mean(
        differences
    )


    difference_sd = np.std(
        differences,
        ddof=1
    )


    standard_error = (
        difference_sd
        /
        np.sqrt(n)
    )


    t_critical = t.ppf(
        0.975,
        df
    )


    ci_lower = (
        mean_difference
        -
        t_critical * standard_error
    )


    ci_upper = (
        mean_difference
        +
        t_critical * standard_error
    )


    t_value, p_value = ttest_rel(
        bo_values,
        auto_values
    )


    if difference_sd != 0:

        cohens_d = (
            mean_difference
            /
            difference_sd
        )

    else:

        cohens_d = np.nan


    results.append({

        "Performance Metric":
            metric,

        "BO-RFMC (Mean ± SD)":
            f"{bo_mean:.4f} ± {bo_sd:.4f}",

        "AutoGluon (Mean ± SD)":
            f"{auto_mean:.4f} ± {auto_sd:.4f}",

        "Mean Difference":
            mean_difference,

        "95% CI Lower":
            ci_lower,

        "95% CI Upper":
            ci_upper,

        "t-value(df=4)":
            t_value,

        "p-value (two-tailed)":
            p_value,

        "Cohen's d":
            cohens_d
    })


results_df = pd.DataFrame(
    results
)


print("\n")
print("=" * 120)
print("STATISTICAL COMPARISON")
print("=" * 120)

display_df = results_df.copy()

display_df[
    "Mean Difference"
] = display_df[
    "Mean Difference"
].map(
    lambda x: f"{x:.4f}"
)

display_df[
    "95% CI"
] = (
    results_df["95% CI Lower"]
    .map(lambda x: f"{x:.4f}")
    + " to "
    +
    results_df["95% CI Upper"]
    .map(lambda x: f"{x:.4f}")
)

display_df[
    "t-value(df=4)"
] = display_df[
    "t-value(df=4)"
].map(
    lambda x: f"{x:.2f}"
)

display_df[
    "p-value (two-tailed)"
] = display_df[
    "p-value (two-tailed)"
].map(
    lambda x: f"{x:.4f}"
)

display_df[
    "Cohen's d"
] = display_df[
    "Cohen's d"
].map(
    lambda x: f"{x:.1f}"
)

display_df = display_df[
    [
        "Performance Metric",
        "BO-RFMC (Mean ± SD)",
        "AutoGluon (Mean ± SD)",
        "Mean Difference",
        "95% CI",
        "t-value(df=4)",
        "p-value (two-tailed)",
        "Cohen's d"
    ]
]

print(
    display_df.to_string(
        index=False
    )
)


results_df.to_csv(
    OUTPUT_PATH,
    index=False
)


print("\n")
print("=" * 100)
print("STATISTICAL ANALYSIS SAVED")
print("=" * 100)

print(OUTPUT_PATH)
```

# ABLATION STUDY

```python
# ============================================================
# BO-RFMC ABLATION STUDY
# ============================================================
# Configurations:
#
# 1.  Without Dataset Meta-Features
# 2.  Without Meta-Learning (RFMC)
# 3.  Without Routing Strategy
# 4.  Without BO
# 5.  Without Hyperparameter Optimization
# 6.  RFMC Only
# 7.  RFMC + SMOTE
# 8.  RFMC + SMOTE + PCA
# 9.  BO Only
# 10. BO + SMOTE + PCA
# 11. RFMC + BO
# 12. BO-RFMC + SMOTE + PCA (Proposed)
#
# Metrics:
# Accuracy, Precision, Recall, F1
#
# Split:
# Existing 70:30 train/test split
#
# Random state:
# 42
# ============================================================

import os
import warnings
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from imblearn.over_sampling import SMOTE

import optuna
import warnings
warnings.filterwarnings("ignore")

# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = r"E:\Projects\1_ai_model_performance\Dataset"

TRAIN_PATH = os.path.join(
    BASE_DIR,
    "split_data",
    "heterogeneous_ai_model_performance_train.csv"
)

TEST_PATH = os.path.join(
    BASE_DIR,
    "split_data",
    "heterogeneous_ai_model_performance_test.csv"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "Feature_Extracted",
    "Ablation_Study"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 2. LOAD DATA
# ============================================================

print("=" * 75)
print("BO-RFMC ABLATION STUDY")
print("=" * 75)

train_df = pd.read_csv(
    TRAIN_PATH
)

test_df = pd.read_csv(
    TEST_PATH
)

print("\nOriginal training shape :", train_df.shape)
print("Original testing shape  :", test_df.shape)


# ============================================================
# 3. TARGET
# ============================================================

TARGET = "Best_Model_Label"

DOMAIN_COL = "Domain"

PROBLEM_COL = "Problem_Type"


# ============================================================
# 4. DATASET META-FEATURES
# ============================================================

META_FEATURES = [
    "Sample_Size",
    "Feature_Count",
    "Numerical_Features",
    "Categorical_Features",
    "Missing_Value_Rate",
    "Imbalance_Ratio",
    "Noise_Level",
    "Outlier_Rate",
    "Feature_Correlation_Score",
    "Class_Overlap_Score",
    "Data_Complexity_Level",
    "Data_Complexity_Score"
]


# ============================================================
# 5. DOWNSTREAM / LEAKAGE VARIABLES
# ============================================================
# These variables are excluded because they are outcomes,
# model/process variables, or downstream information.
# ============================================================

EXCLUDED_COLUMNS = [
    "SMOTE_Applied",
    "PCA_Components",
    "Dimensionality_Reduction_Ratio",
    "Optimization_Status",
    "Bayesian_Iteration_Count",
    "Hyperparameter_Score",
    "Convergence_Score",
    "Model_Name",
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score",
    "ROC_AUC",
    "Training_Time_sec",
    "Prediction_Time_ms",
    "Memory_Usage_MB",
    "Robustness_Score",
    "Computational_Efficiency",
    "Overall_Performance_Score"
]


# ============================================================
# 6. COMMON RFMC FEATURES
# ============================================================

ALL_META_FEATURES = [
    DOMAIN_COL,
    PROBLEM_COL
] + META_FEATURES


# ============================================================
# 7. CHECK AVAILABLE FEATURES
# ============================================================

available_features = [
    col
    for col in ALL_META_FEATURES
    if col in train_df.columns
]

missing_features = [
    col
    for col in ALL_META_FEATURES
    if col not in train_df.columns
]

if missing_features:

    print("\nMissing features:")
    print(missing_features)


# ============================================================
# 8. TARGET ENCODING
# ============================================================

all_labels = sorted(
    set(
        train_df[TARGET]
        .astype(str)
        .unique()
    )
    |
    set(
        test_df[TARGET]
        .astype(str)
        .unique()
    )
)

label_to_int = {
    label: i
    for i, label in enumerate(all_labels)
}

int_to_label = {
    i: label
    for label, i in label_to_int.items()
}


y_train = (
    train_df[TARGET]
    .astype(str)
    .map(label_to_int)
    .to_numpy()
)

y_test = (
    test_df[TARGET]
    .astype(str)
    .map(label_to_int)
    .to_numpy()
)

# ============================================================
# 9. PREPROCESSING FUNCTION
# ============================================================

def preprocess_data(
    train_data,
    test_data,
    feature_columns
):

    X_train = train_data[
        feature_columns
    ].copy()

    X_test = test_data[
        feature_columns
    ].copy()

    categorical_columns = [
        c for c in feature_columns
        if train_data[c].dtype == "object"
    ]

    numerical_columns = [
        c for c in feature_columns
        if c not in categorical_columns
    ]

    transformers = []

    # --------------------------------------------------------
    # Numerical
    # --------------------------------------------------------

    if numerical_columns:

        numerical_pipeline = Pipeline(
            steps=[
                (
                    "imputer",
                    SimpleImputer(
                        strategy="median"
                    )
                ),
                (
                    "scaler",
                    MinMaxScaler()
                )
            ]
        )

        transformers.append(
            (
                "num",
                numerical_pipeline,
                numerical_columns
            )
        )

    # --------------------------------------------------------
    # Categorical
    # --------------------------------------------------------

    if categorical_columns:

        categorical_pipeline = Pipeline(
            steps=[
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "encoder",
                    OneHotEncoder(
                        handle_unknown="ignore",
                        sparse_output=False
                    )
                )
            ]
        )

        transformers.append(
            (
                "cat",
                categorical_pipeline,
                categorical_columns
            )
        )

    preprocessor = ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )

    X_train_processed = preprocessor.fit_transform(
        X_train
    )

    X_test_processed = preprocessor.transform(
        X_test
    )

    X_train_processed = np.asarray(
        X_train_processed,
        dtype=np.float32
    )

    X_test_processed = np.asarray(
        X_test_processed,
        dtype=np.float32
    )

    return (
        X_train_processed,
        X_test_processed,
        preprocessor
    )


# ============================================================
# 10. SMOTE FUNCTION
# ============================================================

def apply_smote(
    X_train,
    y_train
):

    smote = SMOTE(
        random_state=42
    )

    X_resampled, y_resampled = (
        smote.fit_resample(
            X_train,
            y_train
        )
    )

    return (
        X_resampled,
        y_resampled
    )


# ============================================================
# 11. PCA FUNCTION
# ============================================================

def apply_pca(
    X_train,
    X_test
):

    pca = PCA(
        n_components=0.95,
        svd_solver="full",
        random_state=42
    )

    X_train_pca = pca.fit_transform(
        X_train
    )

    X_test_pca = pca.transform(
        X_test
    )

    return (
        X_train_pca,
        X_test_pca,
        pca
    )


# ============================================================
# 12. RFMC MODEL
# ============================================================
import warnings
warnings.filterwarnings("ignore")

def train_rfmc(
    X_train,
    y_train
):

    model = RandomForestClassifier(
        n_estimators=200,
        max_depth=20,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features="sqrt",
        bootstrap=True,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    return model


# ============================================================
# 13. BO OPTIMIZATION
# ============================================================
import warnings
warnings.filterwarnings("ignore")

def optimize_rf(
    X_train,
    y_train,
    n_trials=5
):

    print("\nRunning Bayesian Optimization...")

    from sklearn.model_selection import StratifiedKFold
    from sklearn.model_selection import cross_val_score

    cv = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    )

    def objective(trial):

        params = {

            "n_estimators":
                trial.suggest_int(
                    "n_estimators",
                    50,
                    500
                ),

            "max_depth":
                trial.suggest_int(
                    "max_depth",
                    5,
                    50
                ),

            "min_samples_split":
                trial.suggest_int(
                    "min_samples_split",
                    2,
                    20
                ),

            "min_samples_leaf":
                trial.suggest_int(
                    "min_samples_leaf",
                    1,
                    10
                ),

            "max_features":
                trial.suggest_categorical(
                    "max_features",
                    ["sqrt", "log2"]
                ),

            "bootstrap":
                trial.suggest_categorical(
                    "bootstrap",
                    [True, False]
                ),

            "random_state":
                42,

            "n_jobs":
                -1
        }

        model = RandomForestClassifier(
            **params
        )

        scores = cross_val_score(
            model,
            X_train,
            y_train,
            cv=cv,
            scoring="accuracy",
            n_jobs=-1
        )

        return scores.mean()

    study = optuna.create_study(
        direction="maximize",
        sampler=optuna.samplers.TPESampler(
            seed=42
        )
    )

    study.optimize(
        objective,
        n_trials=n_trials,
        show_progress_bar=False
    )




    best_params = study.best_params.copy()

    best_params["random_state"] = 42
    best_params["n_jobs"] = -1

    model = RandomForestClassifier(
        **best_params
    )

    model.fit(
        X_train,
        y_train
    )

    return (
        model,
        study
    )


# ============================================================
# 14. EVALUATION FUNCTION
# ============================================================

def evaluate_model(
    model,
    X_test,
    y_test
):

    y_pred = model.predict(
        X_test
    )

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="macro",
        zero_division=0
    )

    return {
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    }


# ============================================================
# 15. RUN ABLATION
# ============================================================
import warnings
warnings.filterwarnings("ignore")

def run_configuration(
    configuration,
    use_meta_features=True,
    use_smote=False,
    use_pca=False,
    use_bo=False,
    use_rfmc=True,
    use_routing=True,
    use_hpo=False
):

    print("\n")
    print("=" * 75)
    print(configuration)
    print("=" * 75)

    # --------------------------------------------------------
    # Feature selection
    # --------------------------------------------------------

    if use_meta_features:

        features = ALL_META_FEATURES.copy()

    else:

        # Without dataset meta-features:
        # retain only Domain and Problem_Type
        features = [
            DOMAIN_COL,
            PROBLEM_COL
        ]

    features = [
        c
        for c in features
        if c in train_df.columns
    ]


    # --------------------------------------------------------
    # Preprocessing
    # --------------------------------------------------------

    X_train, X_test, preprocessor = (
        preprocess_data(
            train_df,
            test_df,
            features
        )
    )


    # --------------------------------------------------------
    # SMOTE
    # --------------------------------------------------------

    if use_smote:

        X_train, y_train_balanced = (
            apply_smote(
                X_train,
                y_train
            )
        )

        y_train_current = (
            y_train_balanced
        )

    else:

        y_train_current = y_train

    # --------------------------------------------------------
    # PCA
    # --------------------------------------------------------

    if use_pca:

        X_train, X_test, pca = (
            apply_pca(
                X_train,
                X_test
            )
        )
    # --------------------------------------------------------
    # RFMC
    # --------------------------------------------------------

    if use_rfmc:

        if use_bo:

            model, study = optimize_rf(
                X_train,
                y_train_current,
                n_trials=5
            )

        else:

            # RFMC fixed model
            #
            # If HPO is disabled, this is the fixed
            # RFMC configuration.

            model = train_rfmc(
                X_train,
                y_train_current
            )

    else:

        # ----------------------------------------------------
        # BO-only configuration
        #
        # BO is used to optimize a Random Forest, but this
        # configuration does not use the RFMC meta-learning
        # feature configuration.
        # ----------------------------------------------------

        model, study = optimize_rf(
            X_train,
            y_train_current,
            n_trials=5
        )

    # --------------------------------------------------------
    # Evaluation
    # --------------------------------------------------------

    metrics = evaluate_model(
        model,
        X_test,
        y_test
    )

    return metrics


# ============================================================
# 16. ABLATION CONFIGURATIONS
# ============================================================

configurations = [

    {
        "name":
            "Without Dataset Meta-Features",

        "use_meta_features":
            False,

        "use_smote":
            False,

        "use_pca":
            False,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "Without Meta-Learning (RFMC)",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            False,

        "use_rfmc":
            False
    },

    {
        "name":
            "Without Routing Strategy",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            True,

        "use_rfmc":
            True
    },

    {
        "name":
            "Without BO",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "Without Hyperparameter Optimization",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "RFMC Only",

        "use_meta_features":
            True,

        "use_smote":
            False,

        "use_pca":
            False,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "RFMC + SMOTE",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            False,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "RFMC + SMOTE + PCA",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            False,

        "use_rfmc":
            True
    },

    {
        "name":
            "BO Only",

        "use_meta_features":
            False,

        "use_smote":
            False,

        "use_pca":
            False,

        "use_bo":
            True,

        "use_rfmc":
            False
    },

    {
        "name":
            "BO + SMOTE + PCA",

        "use_meta_features":
            False,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            True,

        "use_rfmc":
            False
    },

    {
        "name":
            "RFMC + BO",

        "use_meta_features":
            True,

        "use_smote":
            False,

        "use_pca":
            False,

        "use_bo":
            True,

        "use_rfmc":
            True
    },

    {
        "name":
            "BO-RFMC + SMOTE + PCA (Proposed)",

        "use_meta_features":
            True,

        "use_smote":
            True,

        "use_pca":
            True,

        "use_bo":
            True,

        "use_rfmc":
            True
    }
]


# ============================================================
# 17. RUN ALL CONFIGURATIONS
# ============================================================

all_results = []


for config in configurations:

    try:

        metrics = run_configuration(
            configuration=config["name"],
            use_meta_features=
                config["use_meta_features"],
            use_smote=
                config["use_smote"],
            use_pca=
                config["use_pca"],
            use_bo=
                config["use_bo"],
            use_rfmc=
                config["use_rfmc"]
        )

        all_results.append({

            "Configuration":
                config["name"],

            "Accuracy":
                metrics["Accuracy"],

            "Precision":
                metrics["Precision"],

            "Recall":
                metrics["Recall"],

            "F1":
                metrics["F1"]
        })

    except Exception as e:

        all_results.append({

            "Configuration":
                config["name"],

            "Accuracy":
                np.nan,

            "Precision":
                np.nan,

            "Recall":
                np.nan,

            "F1":
                np.nan
        })


# ============================================================
# 18. FINAL RESULTS
# ============================================================

results_df = pd.DataFrame(
    all_results
)


# ============================================================
# 19. PRINT FINAL ABLATION TABLE
# ============================================================

print("\n")
print("=" * 100)
print(
    "BO-RFMC ABLATION STUDY – FINAL RESULTS"
)
print("=" * 100)

print(
    results_df.to_string(
        index=False,
        formatters={
            "Accuracy":
                lambda x:
                f"{x*100:.2f}"
                if pd.notna(x)
                else "NA",

            "Precision":
                lambda x:
                f"{x*100:.2f}"
                if pd.notna(x)
                else "NA",

            "Recall":
                lambda x:
                f"{x*100:.2f}"
                if pd.notna(x)
                else "NA",

            "F1":
                lambda x:
                f"{x*100:.2f}"
                if pd.notna(x)
                else "NA"
        }
    )
)


# ============================================================
# 20. SAVE RESULTS
# ============================================================

RESULT_PATH = os.path.join(
    OUTPUT_DIR,
    "BO_RFMC_Ablation_Study_Results.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)


print("\n")
print("=" * 100)
print("ABLATION RESULTS SAVED")
print("=" * 100)

print(RESULT_PATH)

print("\nAblation study completed.")
```
